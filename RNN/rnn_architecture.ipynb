```python
from keras import Sequential
from keras.layers import Input, Dense, SimpleRNN
```

```python
model = Sequential([
    Input(shape=(4,5)),
    SimpleRNN(3),
    Dense(1, activation='sigmoid')
])
model.summary()
```

```python
from keras.utils import plot_model
plot_model(model, show_shapes=True, to_file='simple_rnn.png')
```
